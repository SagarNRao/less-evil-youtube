```python
import pandas as pd
import requests
from dotenv import load_dotenv
import os

# Load environment variables
load_dotenv()

# Load the CSV file
df = pd.read_csv("newDistracting.csv")

# Create an empty DataFrame to store the results
savedf = pd.DataFrame()

# Collect all video IDs
idArray = df['id'].astype(str).tolist()

# API_KEY = os.getenv("API_KEY")
API_KEY = "AIzaSyCmhLp--zBH_ZIwfCKx8prox4qAyfc_Y8U"

output_file = "newDistracting.csv"

savedf.to_csv(output_file, index=False)

for video_id in idArray:
    url = f'https://youtube.googleapis.com/youtube/v3/videos?part=topicDetails,snippet&id={video_id}&key={API_KEY}'
    response = requests.get(url)

    if response.status_code == 200:
        api_data = response.json()
        if 'items' in api_data and len(api_data['items']) > 0:
            item = api_data['items'][0]

            video_data = {
                'video_id': video_id,
                'title': item['snippet']['title'] if 'snippet' in item else None,
                'description': item['snippet']['description'] if 'snippet' in item else None,
                'channel_name': item['snippet']['channelTitle'] if 'snippet' in item else None,
                'channel_id': item['snippet']['channelId'] if 'snippet' in item else None,
                'tags': ', '.join(item['snippet'].get('tags', [])) if 'snippet' in item else None,
                'topic_details': ', '.join(item['topicDetails'].get('topicCategories', [])) if 'topicDetails' in item else None
            }

            video_df = pd.DataFrame([video_data])

            video_df.to_csv(output_file, mode='a', header=False, index=False)
            print(f"Saved data for video ID {video_id}")
        else:
            empty_data = {
                'video_id': video_id,
                'title': None,
                'description': None,
                'channel_name': None,
                'channel_id': None,
                'tags': None,
                'topic_details': None
            }
            empty_df = pd.DataFrame([empty_data])
            empty_df.to_csv(output_file, mode='a', header=False, index=False)
            print(f"No data found for video ID {video_id}")
    else:
        empty_data = {
            'video_id': video_id,
            'title': None,
            'description': None,
            'channel_name': None,
            'channel_id': None,
            'tags': None,
            'topic_details': None
        }
        empty_df = pd.DataFrame([empty_data])
        empty_df.to_csv(output_file, mode='a', header=False, index=False)
        print(f"Failed to fetch data for video ID {video_id}: {response.json()}")

print("Data fetching complete. Results saved to response_data.csv")
```
