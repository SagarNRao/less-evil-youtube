```python
import pandas as pd
import requests

# Load the CSV file
df = pd.read_csv("combined.csv")

# Create an empty DataFrame to store the results
savedf = pd.DataFrame()

# Collect all video IDs
idArray = df['id'].astype(str).tolist()

# Define the API key
API_KEY = "AIzaSyB5ZAgrVpwQPQ9-azlbQ80lMTsGP2kbzmk"

# Output CSV file
output_file = "response_data.csv"

# Create the CSV file with headers if it doesn't exist
savedf.to_csv(output_file, index=False)

# Fetch data for each video ID and save it immediately
for video_id in idArray:
    url = f'https://youtube.googleapis.com/youtube/v3/videos?part=topicDetails,snippet&id={video_id}&key={API_KEY}'
    response = requests.get(url)
    
    if response.status_code == 200:
        api_data = response.json()
        if 'items' in api_data and len(api_data['items']) > 0:
            video_data = pd.json_normalize(api_data['items'])
            
            # Append the video data to the CSV file
            video_data.to_csv(output_file, mode='a', header=False, index=False)
            print(f"Saved data for video ID {video_id}")
        else:
            print(f"No data found for video ID {video_id}")
    else:
        print(f"Failed to fetch data for video ID {video_id}: {response.json()}")
```
